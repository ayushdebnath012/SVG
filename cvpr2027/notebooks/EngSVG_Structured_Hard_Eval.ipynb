```python
import pathlib, subprocess, sys, shutil, torch

def run(args):
    print("RUN:", " ".join(map(str, args[:5])), flush=True)
    p = subprocess.Popen(args, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
    for line in p.stdout:
        print(line, end="", flush=True)
    if p.wait():
        raise RuntimeError(f"command failed: {p.returncode}")

assert torch.cuda.is_available(), "CUDA GPU required"
print("GPU:", torch.cuda.get_device_name(0), flush=True)
root = pathlib.Path("/content/SVG")
if not root.exists():
    run(["git", "clone", "--depth", "1", "--branch", "cvpr2027-research-package", "https://github.com/ayushdebnath012/SVG.git", str(root)])
run([sys.executable, "-m", "pip", "install", "-q", "svgpathtools==1.7.2", "peft", "transformers", "accelerate"])
run([sys.executable, "-m", "pip", "uninstall", "-y", "-q", "torchao"])
scripts = root / "cvpr2027/scripts"
shutil.copy2("/content/structured_engsvg.py", scripts / "structured_engsvg.py")
shutil.copy2("/content/structured_engsvg_hard_eval.py", scripts / "structured_engsvg_hard_eval.py")
run([sys.executable, str(scripts / "structured_engsvg.py"), "train", "--out", "/content/engsvg-structured-run"])
run([sys.executable, str(scripts / "structured_engsvg_hard_eval.py"), "--run", "/content/engsvg-structured-run"])
print("DOWNLOAD: /content/engsvg-structured-run-artifacts.zip", flush=True)
```
